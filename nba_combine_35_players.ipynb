## **Initial 35 Player NBA Data**

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("nba_combine_35_players_updated.csv")

display(data)
```

Scatterplot 1 Height vs. Vertical Jump

```python
print(data.columns.tolist())

graph_data = data.dropna(
    subset=["Height without shoes (in)", "Moving/max vertical (in)"]
)

plt.scatter(
    graph_data["Height without shoes (in)"],
    graph_data["Moving/max vertical (in)"]
)

plt.xlabel("Height Without Shoes (inches)")
plt.ylabel("Moving Vertical Jump (inches)")
plt.title("NBA Player Height vs. Vertical Jump")

plt.show()
```

Scatter Plot 2 Wingspan vs. Moving Vertical

```python
print(data.columns.tolist())

graph_data = data.dropna(
    subset=["Wingspan (in)", "Moving/max vertical (in)"]
)

plt.scatter(
    graph_data["Wingspan (in)"],
    graph_data["Moving/max vertical (in)"]
)

plt.xlabel("Wingspan (inches)")
plt.ylabel("Moving Vertical Jump (inches)")
plt.title("NBA Player Wingspan vs. Vertical Jump")

plt.show()
```

Scatterplot 3: Height vs. Standing Vertical

```python
graph_data = data.dropna(
    subset=["Height without shoes (in)", "Standing vertical (in)"]
)

plt.scatter(
    graph_data["Height without shoes (in)"],
    graph_data["Standing vertical (in)"]
)

plt.xlabel("Height Without Shoes (inches)")
plt.ylabel("Standing Vertical Jump (inches)")
plt.title("NBA Player Height vs. Standing Vertical Jump")

plt.show()
```

Scatter Plot 4: Wingspan vs. Standing Vertical

```python
graph_data = data.dropna(
    subset=["Wingspan (in)", "Standing vertical (in)"]
)

plt.scatter(
    graph_data["Wingspan (in)"],
    graph_data["Standing vertical (in)"]
)

plt.xlabel("Wingspan (inches)")
plt.ylabel("Standing Vertical Jump (inches)")
plt.title("NBA Player Wingspan vs. Standing Vertical Jump")

plt.show()
```

**Correlation Calculations**

```python
height_moving = data["Height without shoes (in)"].corr(
    data["Moving/max vertical (in)"]
)

wingspan_moving = data["Wingspan (in)"].corr(
    data["Moving/max vertical (in)"]
)

height_standing = data["Height without shoes (in)"].corr(
    data["Standing vertical (in)"]
)

wingspan_standing = data["Wingspan (in)"].corr(
    data["Standing vertical (in)"]
)

print("Height vs Moving Vertical:", height_moving)
print("Wingspan vs Moving Vertical:", wingspan_moving)
print("Height vs Standing Vertical:", height_standing)
print("Wingspan vs Standing Vertical:", wingspan_standing)
```

Top 5 Vertical Jumpers

```python
top_jumpers = data.sort_values(
    by="Moving/max vertical (in)",
    ascending=False
)

print(top_jumpers[["Player", "Moving/max vertical (in)"]].head(5))
```

Descriptive Statistics
